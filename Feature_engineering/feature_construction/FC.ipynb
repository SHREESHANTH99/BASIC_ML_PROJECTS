# Feature construction:-feature splitting
Feature Splitting is the process of splitting a feature into multiple features. This is done to capture the underlying structure of the data and to improve the performance of the model.
For example, if we have a feature called "date of birth", we can split it
into three features: "day of birth", "month of birth", and "year of birth". This can help the model to capture the underlying structure of the data and to improve the performance of the model.



```python
import numpy as np 
import pandas as pd 
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
import seaborn as sns
```

```python
df=pd.read_csv('../../feature_engineering/discretization&binarization/train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.head()
```

```python
X=df.drop('Survived',axis=1)
y=df['Survived']
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,cv=19))
```

# Applying feature construction on the titanic dataset

```python
X['Family_size']=X['SibSp']+X['Parch']+1
```

```python
X.head()
```

```python
def myfunc(num):
    if num==1:
        return 0
    elif num>1 and num<=4:
        return 1
    else:
        return 2
```

```python
X['Family_type']=X['Family_size'].apply(myfunc)
```

```python
X.head()
```

```python
X.drop(columns=['SibSp','Parch','Family_size'],inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=19))
```
