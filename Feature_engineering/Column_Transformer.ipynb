```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('./covid_toy.csv')
df['cough'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['has_covid']), df['has_covid'], test_size=0.2)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler,OrdinalEncoder
from sklearn.impute import SimpleImputer
si=SimpleImputer(strategy='most_frequent')
X_Train_fever=si.fit_transform(X_train[['fever']])
X_test_fever=si.transform(X_test[['fever']])

X_Train_fever
```

```python
oe=OrdinalEncoder(categories=[['Mild','Strong']],dtype=int)
X_train_cough=oe.fit_transform(X_train[['cough']])  
X_test_cough=oe.transform(X_test[['cough']])
X_train_cough
```

```python
ohe=OneHotEncoder(sparse_output=False,dtype=int)
X_train_gender_city=ohe.fit_transform(X_train[['gender','city']])
X_test_gender_city=ohe.transform(X_test[['gender','city']])
X_train_gender_city
```

```python
transformer=ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(strategy='most_frequent'),['fever']),
    ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']],dtype=int),['cough']),
    ('tnf3',OneHotEncoder(sparse_output=False,dtype=int),['gender','city'])
],remainder='passthrough')
```

```python
transformer.fit_transform(X_train)
transformer.transform(X_test)
```
