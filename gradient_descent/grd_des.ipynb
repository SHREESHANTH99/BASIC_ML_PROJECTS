```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y=make_regression(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
```

Gradient descent is an optimization algorithm used to minimize a cost function by iteratively moving towards the steepest descent direction. In the context of linear regression, the cost function is typically the mean squared error (MSE) between the predicted values and the actual target values.

```python
y_pred=((78.35*X)+0).reshape(4)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=13)
from sklearn.linear_model import LinearRegression

reg=LinearRegression()
```

```python
plt.scatter(X,y)
plt.plot(X,y_pred,color='red')
plt.plot(X_train,reg.fit(X_train,y_train).predict(X_train),color='green')
```

```python
#proper gradient descent apply:-
b=-100
m=78.35
lr=0.1
epochs=1000
for i in range(epochs):
    loss_slope=-2*np.sum(y-m*X.ravel()-b)
    b=b-(loss_slope*lr)
    y_pred=m*X+b
    plt.plot(X,y_pred)
plt.scatter(X,y)
```
