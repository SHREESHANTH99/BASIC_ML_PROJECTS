## 3. Opening a csv using Url

```python
import requests 
from io import StringIO

url="https://gist.githubusercontent.com/dsternlicht/74020ebfdd91a686d71e785a79b318d4/raw/d3104389ba98a8605f8e641871b9ce71eff73f7e/chartsninja-data-1.csv"
##just telling that request go through this headers
headers={"User-Agents":"Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14;rv:66.0) Gecko/20100101 Firefox/66.0"}
req=requests.get(url,headers=headers)
data =StringIO(req.text)
pd.read_csv(data)
```
