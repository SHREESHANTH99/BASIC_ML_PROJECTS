```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, VotingRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('./boston_house_prices.csv')
df.head()
df.columns=df.iloc[0]
df=df[1:]
df.head()
```

```python
X=df.drop('MEDV',axis=1)
y=df['MEDV']
```

```python
lr=LinearRegression()
dt=DecisionTreeRegressor()
rf=RandomForestRegressor()
svr=SVR()
```

```python
estimators=[('lr',lr),('dt',dt),('rf',rf),('svr',svr)]
```

```python
for estimator in estimators:
    score=cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
    print(f"{estimator[0]}: {np.mean(score)}")
```

```python
from sklearn.ensemble import VotingRegressor
scores=cross_val_score(VotingRegressor(estimators=estimators),X,y,scoring='r2',cv=10)
print(f"Voting Regressor: {np.mean(scores)}")
```

```python
for i in range(1, 5):
    for j in range(1, 5):
        for k in range(1, 5):
            for l in range(1, 5):
                estimators=[('lr',LinearRegression()),('dt',DecisionTreeRegressor(max_depth=i)),('rf',RandomForestRegressor(n_estimators=j)),('svr',SVR(C=k, epsilon=l))]
                scores=cross_val_score(VotingRegressor(estimators=estimators),X,y,scoring='r2',cv=10)
                print(f"Voting Regressor with dt max_depth={i}, rf n_estimators={j}, svr C={k}, svr epsilon={l}: {np.mean(scores)}")
```
